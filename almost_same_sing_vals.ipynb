```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    'text.usetex': True,
    'font.size': 18
})

print('Code to reproduce the results from experiment in Secion 3.1')
print('Example in which our method just needs one step, but power iteration needs a lot')
print('')

# Choose a small epsilon that controls the ration of the two singular values:
epsilon = 1e-2 # The smaller epsilon, the longer power iteration takes

print(f'Consider A = [[1, epsilon],[0,1]] with epsilon = {epsilon}')

# This is our matrix
A = np.array([[1, epsilon],[0, 1]])

# Formula for that largest singular value from exact computations
lambdamax = 1 + (epsilon**2 + epsilon*np.sqrt(epsilon**2+4))/2
normA = np.sqrt(lambdamax)
print(f'Norm of A: {normA}')
print('')

# fix seed for reproducability - different seed will change initialization for
# the power iteration and hence, change the number of iterations needed.
np.random.seed(42342)
# Do our method for the calculation of the norm:
v = np.random.randn(2)
v /= np.linalg.norm(v)
Av = A@v
normAv = np.linalg.norm(Av)

print('')
print('Our method')
for k in range(20): # Just do at most 20 iterations
    y = np.random.randn(2)
    x = y - np.sum(y*v)*v
    x /= np.linalg.norm(x)
    Ax = A@x
    normAx = np.linalg.norm(Ax)
    a = np.sum(Ax*Av)
    b = normAx**2 - normAv**2
    tau = np.sign(a)*(b/(2*np.abs(a)) + np.sqrt(b**2/(4*a**2)+1))
    v += tau*x
    v /= np.linalg.norm(v)
    Av = A@v
    nA = np.linalg.norm(Av)
    print(f'Iter {k+1} norm est: {nA}')
    if np.abs(nA-normA)<1e-5: # stop once we are close enough to the true solution
        print(f'Reached tolerance 10^-5 after {k+1} iterations')
        break


# Now do power iteration
x = np.random.randn(2)
x /= np.linalg.norm(x)

print('')
print('Power iteration')
for k in range(5000): #At most 5000 iterations
    Ax = A@x
    ATAx = A.T@Ax
    nA = np.linalg.norm(Ax)
    print(f'Iter {k+1}, norm est = {nA}')
    if np.abs(nA-normA)<1e-5:
        print(f'Reached tolerance 10^-5 after {k+1} iterations')
        break
    x = ATAx/np.linalg.norm(ATAx)
```
